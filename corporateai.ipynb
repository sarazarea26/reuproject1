```python
#publications only 
pubs = df_yearly[df_yearly["Dataset"] == "Publications"]
pubs
```

```python
#ai publications only 
ai_pubs = pubs[pubs["Category"] == "AI publications"]
ai_pubs
```

```python
#the biggest AI publishers 
top_companies = (ai_pubs.groupby("Name")["Value"].sum().sort_values(ascending=False).head(20))
top_companies
```

```python
#the smallest AI publishers 
bottom_companies = ai_pubs.groupby("Name")["Value"].sum().sort_values().head(50)
bottom_companies
```

```python
#IBM plot
ibm = ai_pubs[ai_pubs["Name"] == "IBM"]

ibm.plot(
    x="Year",
    y="Value",
    kind="line",
    figsize=(8,5)
)
```

```python
microsoft = ai_pubs[ai_pubs["Name"] == "Microsoft"]

microsoft.plot(
    x="Year",
    y="Value",
    kind="line",
    figsize=(8,5)
)
```

```python
meta = ai_pubs[ai_pubs["Name"] == "Meta"]

meta.plot(
    x="Year",
    y="Value",
    kind="line",
    figsize=(8,5)
)
```

```python
top_companies = (ai_pubs.groupby("Name")["Value"].sum().sort_values(ascending=False).head(20).reset_index())
top_companies
```

```python
top_companies.plot.bar(title="Total Number of AI Research Publications Per Company", x="Name", y="Value",
                       grid=True, figsize=(8,4), ylabel="Number of Publications", yticks=[0, 2000, 4000, 6000, 8000, 10000, 12000], xlabel="Companies")
```
